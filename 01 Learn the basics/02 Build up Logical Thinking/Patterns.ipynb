### Pattern 7

```text

*********

````python
def pattern7(n):
    for i in range(n):
        # Space
        for j in range(n - i - 1):
            print(" ", end="")

        # Star
        for j in range(2 * i + 1):
            print("*", end="")

        print()
````

### Pattern 12

```text
1      1
12    21
123  321
12344321

````python
def pattern12(n):
    space = 2*(n-1)
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(j,end="")

        for j in range(space):
            print(" ",end="")

        for j in range(i,0,-1):
            print(j,end="")

        print()
        space -=2
````

### Pattern 19

```text
**********
****  ****
***    ***
**      **
*        *
**      **
***    ***
****  ****
**********

````python
def pattern19(n):
    space = 0

    for i in range(n):
        for j in range(n-i):
            print("*",end='')

        for j in range(space):
            print(" ",end='')

        for j in range(n-i):
            print("*",end='')

        space +=2
        print()

    space = 2 * (n - 2)

    for i in range(2,n+1):
        for j in range(1,i+1):
            print("*",end='')

        for j in range(space):
            print(" ",end='')

        for j in range(1,i+1):
            print("*",end='')

        space -=2
        print()
````

### Pattern 20

```text
*        *
**      **
***    ***
****  ****
**********
****  ****
***    ***
**      **
*        *

````python
def pattern20(n):
    space = 2 * n - 2

    for i in range(2 * n - 1):
        if i < n:
            star = i + 1
        else:
            star = 2 * n - i - 1

        # Left stars
        for j in range(star):
            print("*", end="")

        # Spaces
        for j in range(space):
            print(" ", end="")

        # Right stars
        for j in range(star):
            print("*", end="")

        print()

        if i < n - 1:
            space -= 2
        else:
            space += 2
````

````python
def pattern21(n):
    for i in range(n):
        for j in range(n):
            if i == 0 or j == 0 or i == n-1 or j == n-1:
                print("*",end="")
            else:
                print(" ",end="")
        print()
````

### Pattern 22

```text

4322234
4333334
4444444

````python
def pattern22(n):
    for i in range(2*n-1):
        for j in range(2*n-1):
            top = i
            left = j
            bottom = (2 * n - 2) - i
            right = (2 * n - 2) - j
            print(n-min(min(top,bottom),min(left,right)),end="")

        print()
````
