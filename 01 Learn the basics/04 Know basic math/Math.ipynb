### Count all Digits of a Number

https://takeuforward.org/plus/dsa/problems/count-all-digits-of-a-number?source=strivers-a2z-dsa-track

```python
def countDigit(n):
    if n==0:
        return 1
    count = 0
    while n > 0:
        count += 1
        n = n//10
        # count = int(log10(n)) + 1
    return count


n = int(input("Enter a Number: "))
result = countDigit(n)
print(f"{n} has {result} digits")

# Time Complexity is O(log10(n))
```

# Reverse Digits of A Number

https://leetcode.com/problems/reverse-integer/description/

```python
def reverse(x: int) -> int:
        sign = -1 if x < 0 else 1
        x = abs(x)
        revNumber = 0
        while x > 0:
            lastDigit = x % 10
            x = x // 10
            revNumber = (revNumber * 10) + lastDigit

        revNumber *= sign

        if revNumber < -2**31 or revNumber > 2**31 - 1:
            return 0

        return revNumber

n = int(input("Enter a Number: "))
result = reverse(n)
print(f"Reverse of {n} is {result}")
```

# Palindrome Number

https://leetcode.com/problems/palindrome-number/description/

```python
def palindrome(n) -> bool:
    num = n
    revNumber = 0
    while n > 0:
        lastDigit = n % 10
        n = n // 10
        revNumber = (revNumber * 10) + lastDigit

    if num == revNumber:
        return True
    else:
        return False

n = int(input("Enter a Number: "))
result = palindrome(n)
if result == True:
    print(f"{n} is Palindrome")
else:
    print(f"{n} is not Palindrome")
```

# Check if the Number is Armstrong

https://takeuforward.org/plus/dsa/problems/check-if-the-number-if-armstrong?source=strivers-a2z-dsa-track

```python
def armstrong(n) -> bool:
    sum = 0
    num = n
    while n>0:
        lastDigit = n%10
        n=n//10
        sum = sum + (lastDigit * lastDigit * lastDigit)

    if sum == num:
        return True
    else:
        return False

n = int(input("Enter a Number: "))
result = armstrong(n)
if result == True:
    print(f"{n} is Armstrong Number")
else:
    print(f"{n} is not Armstrong Number")
```

# Divisors of a Number

https://takeuforward.org/plus/dsa/problems/divisors-of-a-number?source=strivers-a2z-dsa-track

```python
from math import sqrt

def printDivisor(n):
    # O(n)
    # for i in range(1,n+1): 
    #     if n%i==0:
    #         print(i, end=",")

    # Optimized:         O(√n)
    # Optimized + sort:  O(√n + k log k)
    num =[]
    for i in range(1, int(sqrt(n)) + 1): # O(√n)
        if n%i==0:
            num.append(i)
            if n // i != i:
                num.append(n//i)

    return sorted(num)

n = int(input("Enter a Number: "))
print(printDivisor(n))
```

# Check for Prime Number

https://takeuforward.org/plus/dsa/problems/check-for-prime-number?source=strivers-a2z-dsa-track

```python
def prime(n):
    count = 0
    for i in range(1, int(sqrt(n)) + 1):
        if n%i==0:
            count +=1
            if n//i != i:
                count +=1

    if count == 2:
        print("Prime")
    else:
        print("Not a Prime")

n = int(input("Enter a Number: "))
prime(n)
```

# GCD of Two Numbers

https://takeuforward.org/plus/dsa/problems/gcd-of-two-numbers?source=strivers-a2z-dsa-track

```python
# O(log(min(a, b)))
def GCD(a : int, b : int):
    while a > 0 and b > 0:
        if a > b:
            a = a%b
        else:
            b = b%a

    if a==0:
        return b

    return a

n1 = int(input("Enter a Number 1: "))
n2 = int(input("Enter a Number 2: "))
result = GCD(n1,n2)
print(f"GCD of {n1} and {n2} is {result}")
```
