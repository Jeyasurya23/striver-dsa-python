# Print 1 to N using Recursion

https://takeuforward.org/plus/dsa/problems/print-1-to-n-using-recursion?source=strivers-a2z-dsa-track

```python
def printNumbers(i,n):
    if i>n:
        return
    print(i)
    printNumbers(i+1,n)

def backtrack(i):
    if i < 1:
        return
    backtrack(i-1)
    print(i)

n = int(input("enter a number:"))
#printNumbers(1,n)
backtrack(n)
```

# Print N to 1 using Recursion

https://takeuforward.org/plus/dsa/problems/print-n-to-1-using-recursion?source=strivers-a2z-dsa-track

```python
def printNumbers(n,i):
    if n==0:
        return
    print(n)
    printNumbers(n-1,i)

def backtrack(i, n):
    if i > n:
        return
    backtrack(i+1, n)
    print(i)

n = int(input("enter a number:"))
#printNumbers(n,1)
backtrack(1,n)
```

# Sum of First N Numbers

https://takeuforward.org/plus/dsa/problems/sum-of-first-n-numbers?source=strivers-a2z-dsa-track

```python
def sumOfNumbers(n:int)->int:
    if(n==0):
        return 0
    
    return n + sumOfNumbers(n-1)

n = int(input("enter a number:"))
sum = sumOfNumbers(n)
print(sum)
```

# Factorial of a given number

https://takeuforward.org/plus/dsa/problems/factorial-of-a-given-number-i?source=strivers-a2z-dsa-track

```python
def factorial(n):
    if(n==0):
        return 1
    return n * factorial(n-1)

n = int(input("enter a number:"))
sum = factorial(n)
print(sum)
```

# Reverse an array

https://takeuforward.org/plus/dsa/problems/reverse-an-array?source=strivers-a2z-dsa-track

```python
def reverse(i: int,arr: list,n: int) -> None:
    if(i>=n//2): return
    arr[i], arr[n-i-1] = arr[n-i-1], arr[i]
    reverse(i+1,arr,n)


n = int(input("Enter a number: "))
arr = []
for i in range(n):
    arr.append(int(input(f"Enter the arr[{i}]: ")))
print(arr)
reverse(0,arr,n)
print(arr)
```
