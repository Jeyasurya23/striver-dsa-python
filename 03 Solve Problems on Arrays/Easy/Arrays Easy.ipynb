### 01. Largest Element - [TUF Link](https://takeuforward.org/plus/dsa/problems/largest-element?source=strivers-a2z-dsa-track)

- **Time Complexity**
    - **Best Case:** O(n)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def largestElement(arr:list[int])->int:
    max = arr[0]
    for a in arr:
        if a > max:
            max = a
    return max

n = int(input("Enter the length of the Array: "))
arr = []
for i in range(n):
    a = int(input(f"Enter the value of index {i}: "))
    arr.append(a)
print(arr)
result = largestElement(arr)
print(f"Largest element in the array is {result}")
```

### 04. Remove duplicates from sorted array - [TUF Link](https://takeuforward.org/plus/dsa/problems/remove-duplicates-from-sorted-array?source=strivers-a2z-dsa-track) | [Leetcode](https://leetcode.com/problems/remove-duplicates-from-sorted-array/description/)

- **Time Complexity**
    - **Best Case:** O(n)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def removeDuplicate(arr:list[int])->int:
    i = 0

    for j in range(1,len(arr)):
        if arr[j] != arr[i]:
            arr[i+1] = arr[j]
            i += 1
    return i+1
            
n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
print(arr)
removeDuplicate(arr)
```

### 05. Left Rotate Array by One - [TUF Link](https://takeuforward.org/plus/dsa/problems/left-rotate-array-by-one?source=strivers-a2z-dsa-track) | [Leetcode](https://leetcode.com/problems/rotate-array/description/)

- **Time Complexity**
    - **Best Case:** O(n)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def leftRotate(arr:list[int])->None:
    temp = arr[0]

    for i in range(1,len(arr)):
        arr[i-1]=arr[i]

    arr[len(arr)-1] = temp

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
print(arr)
leftRotate(arr)
print(arr)
```

### 07. Move Zeros to End - [TUF Link](https://takeuforward.org/plus/dsa/problems/move-zeros-to-end?source=strivers-a2z-dsa-track) | [Leetcode](https://leetcode.com/problems/move-zeroes/description/)

- **Time Complexity**
    - **Best Case:** O(n)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def moveZeroToEnd(arr:list[int])->None:
    j = -1

    for i in range(len(arr)):
        if arr[i] == 0:
            j = i
            break

    if j==-1: return

    for i in range (j+1,len(arr)):
        if arr[i] != 0:
            arr[i],arr[j] = arr[j],arr[i]
            j+=1

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
print(arr)
moveZeroToEnd(arr)
print(arr)
```

### 08. Linear Search - [TUF Link](https://takeuforward.org/plus/dsa/problems/linear-search?source=strivers-a2z-dsa-track)

- **Time Complexity**
    - **Best Case:** O(1)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def linearSearch(arr:list[int],target:int)->int:
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
target = int(input("Enter the number to find in the array: "))
print(arr)
result=linearSearch(arr,target)
print(result)
```

### 11. Find the Missing Number - [TUF Link](https://takeuforward.org/plus/dsa/problems/find-missing-number?source=strivers-a2z-dsa-track) | [Leetcode Link](https://leetcode.com/problems/missing-number/description/)

- **Time Complexity**
    - **Best Case:** O(n)
    - **Average Case:** O(n)
    - **Worst Case:** O(n)
- **Space Complexity**
    - O(1)

```python
def missingNumber(arr: list[int]) -> int:
    xor1 = 0
    xor2 = 0

    for i in range(len(arr)):
        xor2 = xor2 ^ arr[i]
        xor1 = xor1 ^ (i + 1)
    return xor1 ^ xor2

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
print(arr)
result = missingNumber(arr)
print(result)
```

### 12. Maximum Consecutive Ones - [TUF Link](https://takeuforward.org/plus/dsa/problems/maximum-consecutive-ones?source=strivers-a2z-dsa-track) | [Leetcode]( https://leetcode.com/problems/max-consecutive-ones/description/)

- **Time Complexity**
    - **Best Case:** O()
    - **Average Case:** O()
    - **Worst Case:** O()
- **Space Complexity**
    - O()

```python
def maxOnes(arr:list[int])->int:
    maxi=0
    count=0
    for i in range(len(arr)):
        if arr[i] == 1:
            count+=1
            maxi = max(maxi,count)
        else:
            count = 0

    return maxi

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
print(arr)
result = maxOnes(arr)
print(result)
```

### 14. Longest Subarray with given Sum K (Positives) - [TUF Link](https://takeuforward.org/plus/dsa/problems/longest-subarray-with-sum-k?source=strivers-a2z-dsa-track)

- **Time Complexity**
    - **Best Case:** O()
    - **Average Case:** O()
    - **Worst Case:** O()
- **Space Complexity**
    - O()

```python
def longestSubArrayWithSumK(arr: list[int], k: int) -> int:
    left = 0
    total = 0
    maxLen = 0

    for right in range(len(arr)):
        total += arr[right]

        while total > k:
            total -= arr[left]
            left += 1

        if total == k:
            maxLen = max(maxLen, right - left + 1)

    return maxLen

n = int(input("Enter the length of the Array: "))
arr = [int(input(f"Enter the value of index {i}: ")) for i in range(n)]
k = int(input("Enter the sum to find: "))
print(arr)
result = longestSubArrayWithSumK(arr,k)
print(result)
```
