### 01. Largest Element - [TUF Link](https://takeuforward.org/plus/dsa/problems/largest-element?source=strivers-a2z-dsa-track)

- **Time Complexity**
    - **Best Case:** O()
    - **Average Case:** O()
    - **Worst Case:** O()
- **Space Complexity**
    - O()

```python
n = int(input("Enter the length of the Array: "))
arr = []
for i in range(n):
    a = int(input(f"Enter the value of index {i}: "))
    arr.append(a)
print(arr)
```
