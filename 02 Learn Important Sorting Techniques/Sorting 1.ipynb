# Selection Sort

- Select the minimum and swap
- Time complexity - O(n^2)

```python
def selectionSort(arr)->None:
    n = len(arr)
    for i in range (n-1):
        min = i
        for j in range (i,n):
            if arr[j]<arr[min]:
                min = j
        arr[i],arr[min] = arr[min],arr[i]


arr = [52, 9, 36 , 35 , 23, 1]
print(arr)
selectionSort(arr)
print(arr)
```

# Bubble Sort

- Push the max to the last by adjacent swap
- Time complexity 
    - Worst case - O(n^2)
    - Average case - O(n^2)
    - Best case - O(n)

```python
def bubbleSort(arr)-> None:
    n= len(arr)
    for i in range(n-1,0,-1):
        didSwap = 0
        for j in range(i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                didSwap = 1
        if didSwap == 0:
            break

arr = [52, 9, 36 , 35 , 23, 1]
print(arr)
bubbleSort(arr)
print(arr)
```

# Insertion Sort

- Take a element and place it in correct position
- Time Complexity
    - Worst case - O(n^2)
    - Average case - O(n^2)
    - Best case - O(n)

```python
def insertionSort(arr)->None:
    n = len(arr)
    for i in range(n):
        j = i
        while j > 0 and arr[j-1] > arr[j]:
            arr[j],arr[j-1] = arr[j-1],arr[j]
            j -= 1

arr = [52, 9, 36 , 35 , 23, 1]
print(arr)
insertionSort(arr)
print(arr)
```
