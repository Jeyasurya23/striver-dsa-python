# Merge Sort

- Divide and Merge
- Time Complexity
    - Worst - O(nlog2n)
    - Average - O(nlog2n)
    - Best - O(nlog2n)
- Space Complexity - O(n)

```python
def merge(arr,low,mid,high):
    left = low
    right = mid + 1
    temp = []

    while left <= mid and right <= high:
        if arr[left] <= arr[right]:
            temp.append(arr[left])
            left+=1
        else:
            temp.append(arr[right])
            right+=1 

    while left<=mid:
        temp.append(arr[left])
        left+=1

    while right<=high:
        temp.append(arr[right])
        right+=1

    for i in range(low,high+1):
        arr[i]=temp[i-low]

def mergeSort(arr,low,high):
    # Base Condition
    if low == high:
        return

    # Mid value
    mid = (low + high) // 2

    # Divide
    mergeSort(arr,low,mid)
    mergeSort(arr,mid+1,high)

    # Merge
    merge(arr,low,mid,high)
    
arr = [35, 7, 6 , 4 , 8, 2, 44, 56, 100, 3]
print(arr)
mergeSort(arr,0,len(arr)-1)
print(arr)
```

# Quick Sort

- Pick a pivot (commonly the last element, first element, or a random one)
- Partition: rearrange the array so the pivot lands in its correct final sorted position, with smaller elements to its left and larger to its right
- Recursively apply the same process to the left and right sub-arrays (excluding the pivot, since it's already placed correctly)
- Time Complexity
    - Worst - O(n²)
    - Average - O(n log n)
    - Best - O(n log n)
- Space Complexity
    - Worst - O(log n)
    - Average - O(log n)
    - Best - O(n)

```python
def partition(arr,low,high):
    pivot = arr[low]
    i = low + 1
    j = high

    while i <= j:
        while i<=high and arr[i] <= pivot:
            i += 1
        while j>=low+1 and arr[j] > pivot:
            j -= 1
        if i < j:
            arr[i],arr[j]=arr[j],arr[i]

    arr[low],arr[j]=arr[j],arr[low]        
    return j


def quickSort(arr, low, high):
    if low<high:
        pivotIndex  = partition(arr,low,high)
        quickSort(arr,low,pivotIndex - 1 ) 
        quickSort(arr,pivotIndex + 1,high)

arr = [10, 7, 8, 9, 1, 5]
print(arr)
quickSort(arr, 0, len(arr) - 1)
print(arr)
```
