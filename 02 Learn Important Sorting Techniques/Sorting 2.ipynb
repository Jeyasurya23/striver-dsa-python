# Merge Sort

- Divide and Merge
- Time Complexity
    - Worst - O(nlog2n)
    - Average - O(nlog2n)
    - Best - O(nlog2n)
- Space Complexity - O(n)

```python
def merge(arr,low,mid,high):
    left = low
    right = mid + 1
    temp = []

    while left <= mid and right <= high:
        if arr[left] <= arr[right]:
            temp.append(arr[left])
            left+=1
        else:
            temp.append(arr[right])
            right+=1 

    while left<=mid:
        temp.append(arr[left])
        left+=1

    while right<=high:
        temp.append(arr[right])
        right+=1

    for i in range(low,high+1):
        arr[i]=temp[i-low]

def mergeSort(arr,low,high):
    # Base Condition
    if low == high:
        return

    # Mid value
    mid = (low + high) // 2

    # Divide
    mergeSort(arr,low,mid)
    mergeSort(arr,mid+1,high)

    # Merge
    merge(arr,low,mid,high)
    
arr = [35, 7, 6 , 4 , 8, 2, 44, 56, 100, 3]
print(arr)
mergeSort(arr,0,len(arr)-1)
print(arr)
```
